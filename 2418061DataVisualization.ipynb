# 11.Visualisasi 2-Membuat Bar Chart Penerbit dengan peminjaman terbanyak

```python
import matplotlib.pyplot as plt

publisher_count = dataset.groupby('Penerbit')['Judul Buku'].count()
publisher_count = publisher_count.sort_values(ascending=False).head(6)

plt.figure(figsize=(12, 5))
publisher_count.plot(
    kind='bar',
    color='green'
)

plt.title('Penerbit dengan Peminjaman Terbanyak',
          fontsize=18, color='blue')
plt.xlabel('Penerbit', fontsize=12)
plt.ylabel('Jumlah Peminjaman', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.ylim(bottom=0)
plt.grid(axis='y', linestyle='-.', linewidth=0.5)
plt.tight_layout()
plt.show()
```

# 12.Visualisasi 3-Membuat Heatmap Judul buku per program studi

```python
import matplotlib.pyplot as plt
import seaborn as sns

top_books = dataset['Judul Buku'].value_counts().head(10).index
filtered_data = dataset[dataset['Judul Buku'].isin(top_books)]

pivot_data = pd.crosstab(
    filtered_data['Program Studi'],
    filtered_data['Judul Buku']
)

plt.figure(figsize=(14, 6))
sns.heatmap(
    pivot_data,
    annot=True,
    fmt='d',
    cmap='YlGnBu'
)

plt.title('Heatmap Peminjaman Buku per Program Studi',
          fontsize=18, color='blue')
plt.xlabel('Judul Buku')
plt.ylabel('Program Studi')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

# 13.Visualisasi 4-Membuat Pie Chart Kelompok usia buku yang dipinjam

```python
import matplotlib.pyplot as plt

age_count = dataset['Kelompok Usia Buku'].value_counts()

# Warna berbeda untuk setiap irisan
warna = ['red', 'orange']

plt.figure(figsize=(10, 7))

age_count.plot(
    kind='pie',
    colors=warna[:len(age_count)],
    autopct='%1.1f%%',
    startangle=90,
    counterclock=False,
    wedgeprops={'edgecolor': 'white', 'linewidth': 1}
)

plt.title(
    'Peminjaman Berdasarkan Kelompok Usia Buku',
    fontsize=18,
    color='blue',
    pad=20
)

plt.ylabel('')
plt.axis('equal')
plt.tight_layout()
plt.show()
```

# 14.Visualisasi 4-Membuat line chart untuk membandingkan jumlah peminjaman berdasarkan Penerbit dan Kelompok Usia Buku

```python
import pandas as pd
import matplotlib.pyplot as plt

# Memeriksa kolom yang dibutuhkan
required_columns = ['Penerbit', 'Kelompok Usia Buku']
missing_columns = [
    col for col in required_columns
    if col not in dataset.columns
]

if missing_columns:
    print("Kolom tidak ditemukan:", missing_columns)
    print("Kolom yang tersedia:", dataset.columns.tolist())

else:
    # Mengambil data yang dibutuhkan
    data_visualisasi = dataset[required_columns].copy()

    # Menghapus data kosong
    data_visualisasi = data_visualisasi.dropna(
        subset=required_columns
    )

    # Menghitung jumlah peminjaman per penerbit
    # berdasarkan kelompok usia buku
    comparison = pd.crosstab(
        data_visualisasi['Penerbit'],
        data_visualisasi['Kelompok Usia Buku']
    )

    # Mengurutkan penerbit berdasarkan total peminjaman
    comparison = comparison.loc[
        comparison.sum(axis=1)
        .sort_values(ascending=False).index
    ]

    # Mengambil maksimal 10 penerbit teratas
    comparison = comparison.head(10)

    if comparison.empty:
        print("Tidak ada data yang dapat divisualisasikan.")

    else:
        # Membuat line chart
        ax = comparison.plot(
            kind='line',
            figsize=(14, 7),
            marker='o',
            linewidth=2,
            markersize=6
        )

        plt.title(
            'Perbandingan Peminjaman Berdasarkan Penerbit '
            'dan Kelompok Usia Buku',
            fontsize=16,
            pad=20
        )

        plt.xlabel('Penerbit', fontsize=12)
        plt.ylabel('Jumlah Peminjaman', fontsize=12)

        plt.xticks(rotation=45, ha='right')
        plt.ylim(bottom=0)

        plt.grid(
            axis='y',
            linestyle='--',
            linewidth=0.5,
            alpha=0.7
        )

        plt.legend(
            title='Kelompok Usia Buku',
            bbox_to_anchor=(1.02, 1),
            loc='upper left'
        )

        plt.tight_layout()
        plt.show()
```
